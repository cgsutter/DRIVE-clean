```python
import pandas as pd
```

```python
# eventually will need to isolate specific date times that had higher levels BUT will definitely also need non icy instances if we do any model training with it
# step 1, just grab samples from range of dates
# take hours 0000 0800 1600 from each date. 62 dates (jan and mar 2023), so that is 62 dates x 3 hours is ~180 instances


dates_list_jan = [
    f"202301{day:02d}_{t}"
    for day in range(1, 32)
    for t in ["0000", "0800", "1600"]
]

dates_list_mar = [
    f"202303{day:02d}_{t}"
    for day in range(1, 32)
    for t in ["0000", "0800", "1600"]
]

dates_list = dates_list_jan+dates_list_mar

forcsv = pd.DataFrame(dates_list, columns=['date'])
```
