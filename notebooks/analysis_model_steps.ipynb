```python
import pandas as pd
import os
```

CNN

```python

CNN_cat_acc = [] #HERE!!
ctcheck = []

dir = "/home/csutter/DRIVE-clean/operational_models/CNN/preds" # HERE!! adjust for diff steps in model

predcsvs = os.listdir(dir)

trackers = [f"{dir}/{i}" for i in predcsvs]

# print(trackers)

dfslist = []

for t in trackers:
    d = pd.read_csv(t)
    v = d[d["innerPhase"]=="innerVal"]
    dfslist.append(v)

df = pd.concat(dfslist)
print(df.columns)
df["correct"] = df["model_pred"] == df["img_cat"] #HERE!! Depending on col name

print("overall acc")
print(sum(df["correct"])/len(df))

cats = ["dry", "poor_viz", "snow","snow_severe","wet"]
for c in cats:
    catsub = df[df["img_cat"]==c]
    catacc = sum(catsub["correct"])/len(catsub)
    CNN_cat_acc.append(catacc) #HERE!! according to list name
    ctcheck.append(len(catsub))

print(f"acc by cat: {cats}")
print(CNN_cat_acc) #HERE!!
print(sum(ctcheck))
```
