```python
import pandas as pd
import os
```

```python
dir = "/home/csutter/DRIVE-clean/operational_models/calibration/calib_downstream_data"

predcsvs = os.listdir(dir)

trackers = [f"{dir}/{i}" for i in predcsvs]

# print(trackers)
```

```python
dfslist = []

for t in trackers:
    d = pd.read_csv(t)
    v = d[d["innerPhase"]=="innerVal"]
    print(len(v))
    dfslist.append(v)

df = pd.concat(dfslist)
df["correct"] = df["calib_pred"] == df["img_cat"]

display(df.head(4))

print(sum(df["correct"])/len(df))
```
