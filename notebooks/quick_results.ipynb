```python
import pandas as pd
import os
```

```python
# Operational models, more data, trained without outerTest
# For paper, we use the uncalibrated preds, which do still exist cleanly in the calibration csv predictions, just in o_pred col

dir = "/home/csutter/DRIVE-clean/calibration/calib_downstreamFinal_data"

# f"/home/csutter/DRIVE-clean/downstream/data_predsFinal"

#  "/home/csutter/DRIVE-clean/operational_models/calibration/calib_downstream_data"

# "/home/csutter/DRIVE-clean/calibration/calib_CNNFinal_data"

# f"/home/csutter/DRIVE-clean/downstream/data_predsFinal" # 80.7

predcsvs = os.listdir(dir)

trackers = [f"{dir}/{i}" for i in predcsvs]

# print(trackers)
```

```python
dfslist = []

for t in trackers:
    d = pd.read_csv(t)
    v = d[d["innerPhase"]=="innerVal"]
    print(len(v))
    dfslist.append(v)

df = pd.concat(dfslist)
df["correct"] = df["o_pred"] == df["img_cat"] # calib_pred for calib, or ds_pred for downstream, o_pred for uncalib CNN (which do live in the calib preds just for reference)

display(df.head(4))

print(sum(df["correct"])/len(df))
```
