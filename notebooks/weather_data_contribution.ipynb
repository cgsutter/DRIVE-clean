```python
# ==========================================
# 3. EXTRACT TRUE PROBS & MERGE
# ==========================================
def get_true_prob_cnn(row):
    return row[f"prob_{row['img_cat']}"]

def get_true_prob_ds(row):
    return row[f"ds_prob_{row['img_cat']}"]

cnn['true_class_prob_cnn'] = cnn.apply(get_true_prob_cnn, axis=1)
ds['true_class_prob_ds'] = ds.apply(get_true_prob_ds, axis=1)

# Merge the explicit dataframes
merged = cnn.merge(
    ds, 
    on=['img_orig', 'tracker', 'img_cat'], 
    how='inner'
)

# Calculate the true probability shift
merged['true_prob_shift'] = merged['true_class_prob_ds'] - merged['true_class_prob_cnn']
print(f"Merged rows: {len(merged)}")


# ==========================================
# 4. SETUP CORRECTNESS FLAGS
# ==========================================
# Now we use the clean column names we established with .rename()
merged['CNN_Correct'] = merged['cnn_pred'] == merged['img_cat']
merged['DS_Correct'] = merged['ds_pred'] == merged['img_cat']

merged['CNN_Result'] = merged['CNN_Correct'].map({True: 'CNN: Correct', False: 'CNN: Incorrect'})
merged['DS_Result'] = merged['DS_Correct'].map({True: 'DS: Correct', False: 'DS: Incorrect'})

row_order = ['CNN: Correct', 'CNN: Incorrect']
col_order = ['DS: Correct', 'DS: Incorrect']


# ==========================================
# 5. MACRO-LEVEL SUMMARIES
# ==========================================

# --- Matrix 1: Counts & Percentages ---
total_obs = len(merged)

def format_counts(x):
    cnt = len(x)
    pct = (cnt / total_obs) * 100
    return f"{cnt} ({pct:.1f}%)"

matrix_1 = merged.groupby(['CNN_Result', 'DS_Result']).apply(format_counts).unstack()
matrix_1 = matrix_1.reindex(index=row_order, columns=col_order)

print("\n=== Matrix 1: Total Observations & Percentages ===")
display(matrix_1)

# --- Alt Matrix 1: Counts & Row Percentages (CNN Baseline) ---
# First, get the raw counts in the correct order

print("When the CNN gets it right, how often does the downstream model preserve that correct answer? When the CNN gets it wrong, how often does the downstream model fix it?")
counts_df = merged.groupby(['CNN_Result', 'DS_Result']).size().unstack(fill_value=0)
counts_df = counts_df.reindex(index=row_order, columns=col_order)

# Calculate the total observations per row (CNN total correct, CNN total incorrect)
row_totals = counts_df.sum(axis=1)

# Define the new column order including the Total column
alt_col_order = ['DS: Correct', 'DS: Incorrect', 'Total (Row)']

# Create an empty dataframe to hold our formatted strings
alt_matrix_1 = pd.DataFrame(index=row_order, columns=alt_col_order)

for row in row_order:
    total = row_totals[row]
    
    # Fill in the standard columns
    for col in col_order:
        count = counts_df.loc[row, col]
        pct = (count / total) * 100 if total > 0 else 0
        alt_matrix_1.loc[row, col] = f"{count} ({pct:.1f}%)"
    
    # Add the explicit Total column at the end
    alt_matrix_1.loc[row, 'Total (Row)'] = f"{total} (100.0%)"

print("\n=== Alt Matrix 1: Counts & ROW Percentages ===")
print("(Percentages sum to 100% horizontally, showing DS performance based on CNN starting point)")
display(alt_matrix_1)

# --- Matrix 2: Average Predicted Probabilities ---
def format_pred_probs(x):
    # Using our cleanly renamed columns here
    cnn_mean = x['cnn_prob'].mean()
    ds_mean = x['ds_prob'].mean()
    return f"CNN: {cnn_mean:.3f} | DS: {ds_mean:.3f}"

matrix_2 = merged.groupby(['CNN_Result', 'DS_Result']).apply(format_pred_probs).unstack()
matrix_2 = matrix_2.reindex(index=row_order, columns=col_order)

print("\n=== Matrix 2: Average Predicted Probabilities ===")
print("(The probability the model assigned to its CHOSEN answer)")
display(matrix_2)


# --- Matrix 3: Average Labeled (True Class) Probabilities ---
def format_true_probs(x):
    cnn_mean = x['true_class_prob_cnn'].mean()
    ds_mean = x['true_class_prob_ds'].mean()
    return f"CNN: {cnn_mean:.3f} | DS: {ds_mean:.3f}"

matrix_3 = merged.groupby(['CNN_Result', 'DS_Result']).apply(format_true_probs).unstack()
matrix_3 = matrix_3.reindex(index=row_order, columns=col_order)

print("\n=== Matrix 3: Average Labeled (True Class) Probabilities ===")
print("(The probability the model assigned to the CORRECT answer)")
display(matrix_3)


# ==========================================
# 6. CLASS-LEVEL SUMMARIES
# ==========================================

# --- Determine Transition States ---
def determine_transition(row):
    if not row['CNN_Correct'] and row['DS_Correct']:
        return '1_Rescued (CNN Wrong -> DS Right)'
    elif row['CNN_Correct'] and not row['DS_Correct']:
        return '2_Ruined (CNN Right -> DS Wrong)'
    elif row['CNN_Correct'] and row['DS_Correct']:
        return '3_Agreed (Correct)'
    else:
        return '4_Agreed (Incorrect)'

merged['transition_state'] = merged.apply(determine_transition, axis=1)

# --- Matrix 4: Rescued vs. Ruined Impact by Class ---
base_counts = merged.groupby(['img_cat', 'transition_state']).size().unstack(fill_value=0)
rescue_col = '1_Rescued (CNN Wrong -> DS Right)'
ruin_col = '2_Ruined (CNN Right -> DS Wrong)'

for col in [rescue_col, ruin_col]:
    if col not in base_counts.columns:
        base_counts[col] = 0

class_totals = base_counts.sum(axis=1)
changed_totals = base_counts[rescue_col] + base_counts[ruin_col]
changed_totals_safe = changed_totals.replace(0, np.nan)

rescue_pct_class = (base_counts[rescue_col] / class_totals) * 100
ruin_pct_class = (base_counts[ruin_col] / class_totals) * 100
rescue_pct_change = (base_counts[rescue_col] / changed_totals_safe) * 100
ruin_pct_change = (base_counts[ruin_col] / changed_totals_safe) * 100

impact_matrix = pd.DataFrame({
    'Total Class Images': class_totals,
    'Rescued Count': base_counts[rescue_col],
    'Ruined Count': base_counts[ruin_col],
    '% of Class Rescued': rescue_pct_class.map('{:.1f}%'.format),
    '% of Class Ruined': ruin_pct_class.map('{:.1f}%'.format),
    'Change Win Rate (Rescued)': rescue_pct_change.fillna(0).map('{:.1f}%'.format),
    'Change Loss Rate (Ruined)': ruin_pct_change.fillna(0).map('{:.1f}%'.format)
})

print("\n=== Matrix 4: Rescued vs. Ruined Impact by Class ===")
display(impact_matrix)

# ==========================================
# 7. SPECIFIC DOMAIN: CNN INCORRECT ONLY
# ==========================================

# --- Matrix 5: Class-Level Rescue Rate ---
# Filter the domain to ONLY cases where the CNN guessed incorrectly
cnn_wrong_df = merged[merged['CNN_Correct'] == False]

# Get the raw counts grouped by image category (rows) and DS result (columns)
counts_df = cnn_wrong_df.groupby(['img_cat', 'DS_Result']).size().unstack(fill_value=0)

# Ensure both DS columns exist just in case a category had 0 for one of them
for col in ['DS: Correct', 'DS: Incorrect']:
    if col not in counts_df.columns:
        counts_df[col] = 0

# Calculate the total CNN errors for each category
row_totals = counts_df.sum(axis=1)

# Define the display columns
display_cols = ['DS: Correct (Rescued)', 'DS: Incorrect (Still Wrong)', 'Total CNN Errors (Row)']

# Create an empty dataframe to hold our formatted strings
matrix_5 = pd.DataFrame(index=counts_df.index, columns=display_cols)

for row in counts_df.index:
    total = row_totals[row]
    
    # Format DS: Correct (The Rescue Rate)
    count_correct = counts_df.loc[row, 'DS: Correct']
    pct_correct = (count_correct / total) * 100 if total > 0 else 0
    matrix_5.loc[row, 'DS: Correct (Rescued)'] = f"{count_correct} ({pct_correct:.1f}%)"
    
    # Format DS: Incorrect (The Unfixed Errors)
    count_incorrect = counts_df.loc[row, 'DS: Incorrect']
    pct_incorrect = (count_incorrect / total) * 100 if total > 0 else 0
    matrix_5.loc[row, 'DS: Incorrect (Still Wrong)'] = f"{count_incorrect} ({pct_incorrect:.1f}%)"
    
    # Add the explicit Total column at the end
    matrix_5.loc[row, 'Total CNN Errors (Row)'] = f"{total} (100.0%)"

print("\n=== Matrix 5: Class-Level Rescue Rate (Domain: CNN Incorrect Only) ===")
print("(Percentages sum to 100% horizontally. Shows exactly what happened to the errors the CNN made.)")
display(matrix_5)

# ==========================================
# 8. SPECIFIC DOMAIN: CNN CORRECT ONLY
# ==========================================

# --- Matrix 6: Class-Level Ruin Rate ---
# Filter the domain to ONLY cases where the CNN guessed correctly
cnn_correct_df = merged[merged['CNN_Correct'] == True]

# Get the raw counts grouped by image category (rows) and DS result (columns)
counts_df_correct = cnn_correct_df.groupby(['img_cat', 'DS_Result']).size().unstack(fill_value=0)

# Ensure both DS columns exist just in case a category had 0 for one of them
for col in ['DS: Correct', 'DS: Incorrect']:
    if col not in counts_df_correct.columns:
        counts_df_correct[col] = 0

# Calculate the total CNN correct predictions for each category
row_totals_correct = counts_df_correct.sum(axis=1)

# Define the display columns
display_cols_correct = ['DS: Correct (Preserved)', 'DS: Incorrect (Ruined)', 'Total CNN Correct (Row)']

# Create an empty dataframe to hold our formatted strings
matrix_6 = pd.DataFrame(index=counts_df_correct.index, columns=display_cols_correct)

for row in counts_df_correct.index:
    total = row_totals_correct[row]
    
    # Format DS: Correct (The Preserved Rate)
    count_correct = counts_df_correct.loc[row, 'DS: Correct']
    pct_correct = (count_correct / total) * 100 if total > 0 else 0
    matrix_6.loc[row, 'DS: Correct (Preserved)'] = f"{count_correct} ({pct_correct:.1f}%)"
    
    # Format DS: Incorrect (The Ruin Rate)
    count_incorrect = counts_df_correct.loc[row, 'DS: Incorrect']
    pct_incorrect = (count_incorrect / total) * 100 if total > 0 else 0
    matrix_6.loc[row, 'DS: Incorrect (Ruined)'] = f"{count_incorrect} ({pct_incorrect:.1f}%)"
    
    # Add the explicit Total column at the end
    matrix_6.loc[row, 'Total CNN Correct (Row)'] = f"{total} (100.0%)"

print("\n=== Matrix 6: Class-Level Ruin Rate (Domain: CNN Correct Only) ===")
print("(Percentages sum to 100% horizontally. Shows exactly what happened to the accurate CNN predictions.)")
display(matrix_6)


# ==========================================
# 9. OVERALL TRANSITION IMPACT BY CLASS
# ==========================================

# --- Matrix 7: Full Transition Breakdown by Class ---
# Get the raw counts grouped by image category and transition state
counts_df_all = merged.groupby(['img_cat', 'transition_state']).size().unstack(fill_value=0)

# Map our technical transition state names to the clean names you requested
col_mapping = {
    '3_Agreed (Correct)': 'Unchanged (Correct)',
    '4_Agreed (Incorrect)': 'Unchanged (Incorrect)',
    '1_Rescued (CNN Wrong -> DS Right)': 'Rescued',
    '2_Ruined (CNN Right -> DS Wrong)': 'Ruined'
}

# Rename the columns in the counts dataframe
counts_df_all = counts_df_all.rename(columns=col_mapping)

# Ensure all 4 columns exist even if a category had 0 in one of them
expected_cols = ['Unchanged (Correct)', 'Unchanged (Incorrect)', 'Rescued', 'Ruined']
for col in expected_cols:
    if col not in counts_df_all.columns:
        counts_df_all[col] = 0

# Reorder columns to exactly match your request
counts_df_all = counts_df_all[expected_cols]

# Calculate the total images for each category (row totals)
row_totals_all = counts_df_all.sum(axis=1)

# Create an empty dataframe to hold our formatted strings
matrix_7 = pd.DataFrame(index=counts_df_all.index)

for row in counts_df_all.index:
    total = row_totals_all[row]
    
    # Format each of the 4 columns
    for col in expected_cols:
        count = counts_df_all.loc[row, col]
        pct = (count / total) * 100 if total > 0 else 0
        matrix_7.loc[row, col] = f"{count} ({pct:.1f}%)"
    
    # Add the explicit Total column at the end
    matrix_7.loc[row, 'Total Images (Row)'] = f"{total} (100.0%)"

print("\n=== Matrix 7: Full Transition Breakdown by Class ===")
print("(Percentages sum to 100% horizontally. Shows exactly what happened to EVERY image in the category.)")
display(matrix_7)
```

```python



# --- Analysis 1: The Transition Matrix (Rescue vs. Ruin) ---
def determine_transition(row):
    cnn_correct = row['model_pred_cnn'] == row['img_cat']
    ens_correct = row['ens_pred'] == row['img_cat']
    
    if not cnn_correct and ens_correct:
        return '1_Rescued (CNN Wrong -> DS Right)'
    elif cnn_correct and not ens_correct:
        return '2_Ruined (CNN Right -> DS Wrong)'
    elif cnn_correct and ens_correct:
        return '3_Agreed (Correct)'
    else:
        return '4_Agreed (Incorrect)'

mergedf['transition_state'] = mergedf.apply(determine_transition, axis=1)

transition_summary = mergedf.groupby(['img_cat', 'transition_state']).size().unstack(fill_value=0)
if not transition_summary.empty:
    transition_summary['Total_Images'] = transition_summary.sum(axis=1)

print("--- Analysis 1: Transition Matrix (Rescue vs. Ruin) ---")
display(transition_summary)


# --- Analysis 2: The Confidence Shift (When Models Agree) ---
agreed_mask = mergedf['model_pred_cnn'] == mergedf['ens_pred']
agreed_df = mergedf[agreed_mask].copy()

# Shift in the probability of the *predicted* class
agreed_df['pred_prob_shift'] = agreed_df['ens_prob'] - agreed_df['model_prob']
agreed_df['is_correct'] = agreed_df['model_pred_cnn'] == agreed_df['img_cat']

confidence_summary = agreed_df.groupby(['img_cat', 'is_correct'])['pred_prob_shift'].agg(
    Avg_Pred_Prob_Change='mean', 
    Observation_Count='count'
).reset_index()

print("\n--- Analysis 2: Predicted Confidence Shift (When Models Agree) ---")
display(confidence_summary)


# --- Analysis 3: Meteorological Dominance (Ruin Patterns) ---
ruined_df = mergedf[mergedf['transition_state'] == '2_Ruined (CNN Right -> DS Wrong)']

if not ruined_df.empty:
    ruin_patterns = ruined_df.groupby(['img_cat', 'ens_pred']).size().reset_index(name='False_Prediction_Count')
    ruin_patterns = ruin_patterns.sort_values(by=['img_cat', 'False_Prediction_Count'], ascending=[True, False])
    
    print("\n--- Analysis 3: Ruin Patterns (What did the DS model guess instead?) ---")
    display(ruin_patterns)
else:
    print("\n--- Analysis 3: No 'Ruined' predictions found! ---")


# --- Analysis 4: True Class Probability Shift (The "Hidden Lift") ---
# This looks at how the probability of the *correct* answer changed, 
# grouped by category and how the models ultimately behaved.
true_prob_summary = mergedf.groupby(['img_cat', 'transition_state'])['true_prob_shift'].agg(
    Avg_True_Prob_Shift='mean',
    Median_True_Prob_Shift='median'
).reset_index()

# Pivot for easier reading: Rows = Categories, Columns = Transition States
true_prob_pivot = true_prob_summary.pivot(
    index='img_cat', 
    columns='transition_state', 
    values='Avg_True_Prob_Shift'
).fillna(0)

print("\n--- Analysis 4: Average Shift in True Class Probability ---")
print("(Positive numbers mean the DS model gave higher probability to the correct answer than the CNN did)")
display(true_prob_pivot)

# --- Analysis 5: Decisiveness Shift (The "Rescue" Margin) ---
# Isolate the cases where the CNN was wrong, but the weather data fixed it
rescued_df = mergedf[mergedf['transition_state'] == '1_Rescued (CNN Wrong -> DS Right)'].copy()

if not rescued_df.empty:
    # Calculate the difference between the DS model's confident correct guess 
    # and the CNN's confident incorrect guess
    rescued_df['winning_prob_diff'] = rescued_df['ens_prob'] - rescued_df['model_prob']
    
    # Group by the true road condition category
    rescue_margin_summary = rescued_df.groupby('img_cat')['winning_prob_diff'].agg(
        Avg_Winning_Prob_Diff='mean',
        Median_Winning_Prob_Diff='median',
        Min_Diff='min', # Shows cases where DS was less confident in the right answer than CNN was in the wrong one
        Max_Diff='max', # Shows massive jumps in confidence
        Rescue_Count='count'
    ).reset_index()
    
    print("\n--- Analysis 5: Decisiveness Shift (The 'Rescue' Margin) ---")
    print("(Positive numbers mean the DS model was MORE confident in the RIGHT answer than the CNN was in the WRONG answer)")
    display(rescue_margin_summary)
else:
    print("\n--- Analysis 5: No 'Rescued' predictions found to analyze! ---")
```

```python
# --- Step 0: Data Preparation ---
# Assuming your merged dataframe is called 'df'
df_ot = mergedf.copy()

# Helper functions to grab the probability of the true labeled class
def get_true_prob_cnn(row):
    return row[f"prob_{row['img_cat']}"]

def get_true_prob_ens(row):
    return row[f"ensembleAvg_{row['img_cat']}"]

# Extract true class probabilities
df_ot['true_class_prob_cnn'] = df_ot.apply(get_true_prob_cnn, axis=1)
df_ot['true_class_prob_ens'] = df_ot.apply(get_true_prob_ens, axis=1)

# Define Correctness
df_ot['CNN_Correct'] = df_ot['model_pred'] == df_ot['img_cat']
df_ot['Ens_Correct'] = df_ot['ens_pred'] == df_ot['img_cat']

# Map booleans to explicit strings for matrix labels
df_ot['CNN_Result'] = df_ot['CNN_Correct'].map({True: 'CNN: Correct', False: 'CNN: Incorrect'})
df_ot['Ens_Result'] = df_ot['Ens_Correct'].map({True: 'Ens: Correct', False: 'Ens: Incorrect'})

row_order = ['CNN: Correct', 'CNN: Incorrect']
col_order = ['Ens: Correct', 'Ens: Incorrect']

# ==========================================
# MACRO-LEVEL SUMMARIES
# ==========================================

# --- Matrix 1: Counts & Percentages ---
total_obs = len(df_ot)

def format_counts(x):
    cnt = len(x)
    pct = (cnt / total_obs) * 100
    return f"{cnt} ({pct:.1f}%)"

matrix_1 = df_ot.groupby(['CNN_Result', 'Ens_Result']).apply(format_counts).unstack()
matrix_1 = matrix_1.reindex(index=row_order, columns=col_order)

print("=== Matrix 1: Total Observations & Percentages ===")
display(matrix_1)

# --- Matrix 2: Average Predicted Probabilities ---
def format_pred_probs(x):
    cnn_mean = x['model_prob'].mean()
    ens_mean = x['ens_prob'].mean()
    return f"CNN: {cnn_mean:.3f} | Ens: {ens_mean:.3f}"

matrix_2 = df_ot.groupby(['CNN_Result', 'Ens_Result']).apply(format_pred_probs).unstack()
matrix_2 = matrix_2.reindex(index=row_order, columns=col_order)

print("\n=== Matrix 2: Average Predicted Probabilities (Probability of the CHOSEN answer) ===")
display(matrix_2)

# --- Matrix 3: Average Labeled (True Class) Probabilities ---
def format_true_probs(x):
    cnn_mean = x['true_class_prob_cnn'].mean()
    ens_mean = x['true_class_prob_ens'].mean()
    return f"CNN: {cnn_mean:.3f} | Ens: {ens_mean:.3f}"

matrix_3 = df_ot.groupby(['CNN_Result', 'Ens_Result']).apply(format_true_probs).unstack()
matrix_3 = matrix_3.reindex(index=row_order, columns=col_order)

print("\n=== Matrix 3: Average True Class Probabilities (Probability of the CORRECT answer) ===")
display(matrix_3)


# ==========================================
# CLASS-LEVEL SUMMARIES
# ==========================================

# --- Determine Transition States ---
def determine_transition(row):
    if not row['CNN_Correct'] and row['Ens_Correct']:
        return '1_Rescued (CNN Wrong -> Ens Right)'
    elif row['CNN_Correct'] and not row['Ens_Correct']:
        return '2_Ruined (CNN Right -> Ens Wrong)'
    elif row['CNN_Correct'] and row['Ens_Correct']:
        return '3_Agreed (Correct)'
    else:
        return '4_Agreed (Incorrect)'

df_ot['transition_state'] = df_ot.apply(determine_transition, axis=1)

# --- Matrix 4: Rescued vs. Ruined Impact by Class ---
base_counts = df_ot.groupby(['img_cat', 'transition_state']).size().unstack(fill_value=0)
rescue_col = '1_Rescued (CNN Wrong -> Ens Right)'
ruin_col = '2_Ruined (CNN Right -> Ens Wrong)'

# Ensure columns exist even if there were 0 rescues or ruins
for col in [rescue_col, ruin_col]:
    if col not in base_counts.columns:
        base_counts[col] = 0

class_totals = base_counts.sum(axis=1)
changed_totals = base_counts[rescue_col] + base_counts[ruin_col]

# Avoid division by zero by replacing 0s with NaNs temporarily
changed_totals_safe = changed_totals.replace(0, np.nan)

rescue_pct_class = (base_counts[rescue_col] / class_totals) * 100
ruin_pct_class = (base_counts[ruin_col] / class_totals) * 100
rescue_pct_change = (base_counts[rescue_col] / changed_totals_safe) * 100
ruin_pct_change = (base_counts[ruin_col] / changed_totals_safe) * 100

impact_matrix = pd.DataFrame({
    'Total Class Images': class_totals,
    'Rescued Count': base_counts[rescue_col],
    'Ruined Count': base_counts[ruin_col],
    '% of Class Rescued': rescue_pct_class.map('{:.1f}%'.format),
    '% of Class Ruined': ruin_pct_class.map('{:.1f}%'.format),
    'Change Win Rate (Rescued)': rescue_pct_change.fillna(0).map('{:.1f}%'.format),
    'Change Loss Rate (Ruined)': ruin_pct_change.fillna(0).map('{:.1f}%'.format)
})

print("\n=== Matrix 4: Rescued vs. Ruined Impact by Class ===")
display(impact_matrix)


# ==========================================
# PROBABILITY SHIFT ANALYSES
# ==========================================

# --- Analysis: Confidence Shift (When Models Agree) ---
agreed_mask = df_ot['model_pred'] == df_ot['ens_pred']
agreed_df = df_ot[agreed_mask].copy()

# Shift in probability for the agreed-upon predicted class
agreed_df['pred_prob_shift'] = agreed_df['ens_prob'] - agreed_df['model_prob']
agreed_df['is_correct'] = agreed_df['model_pred'] == agreed_df['img_cat']

confidence_summary = agreed_df.groupby(['img_cat', 'is_correct'])['pred_prob_shift'].agg(
    Avg_Pred_Prob_Change='mean', 
    Observation_Count='count'
).reset_index()

print("\n--- Predicted Confidence Shift (When Models Agree) ---")
display(confidence_summary)

# --- Analysis: Decisiveness Shift (The "Rescue" Margin) ---
rescued_df = df_ot[df_ot['transition_state'] == rescue_col].copy()

if not rescued_df.empty:
    rescued_df['winning_prob_diff'] = rescued_df['ens_prob'] - rescued_df['model_prob']
    
    rescue_margin_summary = rescued_df.groupby('img_cat')['winning_prob_diff'].agg(
        Avg_Winning_Prob_Diff='mean',
        Min_Diff='min',
        Max_Diff='max',
        Rescue_Count='count'
    ).reset_index()
    
    print("\n--- Decisiveness Shift (The 'Rescue' Margin) ---")
    display(rescue_margin_summary)
else:
    print("\n--- No 'Rescued' predictions found to analyze! ---")

# --- Analysis: Average Shift in True Class Probability ---
df_ot['true_prob_shift'] = df_ot['true_class_prob_ens'] - df_ot['true_class_prob_cnn']

true_prob_summary = df_ot.groupby(['img_cat', 'transition_state'])['true_prob_shift'].agg(
    Avg_True_Prob_Shift='mean'
).reset_index()

true_prob_pivot = true_prob_summary.pivot(
    index='img_cat', 
    columns='transition_state', 
    values='Avg_True_Prob_Shift'
).fillna(0)

print("\n--- Average Shift in True Class Probability ---")
display(true_prob_pivot)
```
