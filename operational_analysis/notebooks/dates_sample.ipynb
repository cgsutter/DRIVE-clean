```python
# eventually will need to isolate specific date times that had higher levels BUT will definitely also need non icy instances if we do any model training with it
# step 1, just grab samples from range of dates
# take hours 0000 0800 1600 from each date. 62 dates (jan and mar 2023), so that is 62 dates x 3 hours is ~180 instances
# in sets 9 through 12 did: "0000", "0800", "1600"
# in sets 13 through 16 did: "0400", "1200", "2000"

# year_ND = "2021"
year_JFM = "2022"

allmos = []

# Nov
# dates_list_nov = [
#     f"{year_ND}11{day:02d}_{t}"
#     for day in range(1, 31)
#     for t in ["0400", "1200", "2000"] 
# ]
# allmos.append(dates_list_nov)

# # Dec
# dates_list_dec = [
#     f"{year_ND}12{day:02d}_{t}"
#     for day in range(1, 32)
#     for t in ["0400", "1200", "2000"]
# ]
# allmos.append(dates_list_dec)

# Jan
dates_list_jan = [
    f"{year_JFM}01{day:02d}_{t}"
    for day in range(1, 32)
    for t in ["0400", "1200", "2000"]
]
allmos.append(dates_list_jan)

# Feb  - ADJUST FOR LEAP YEAR AS NEEDED
dates_list_feb = [
    f"{year_JFM}02{day:02d}_{t}"
    for day in range(1, 29) # 29, o/w would be range 30 for leap years
    for t in ["0400", "1200", "2000"]
]
allmos.append(dates_list_feb)

# Mar
dates_list_mar = [
    f"{year_JFM}03{day:02d}_{t}"
    for day in range(1, 32)
    for t in ["0400", "1200", "2000"]
]
allmos.append(dates_list_mar)


dates_list = []
for m in allmos:
    for d in m:
        dates_list.append(d)


```
