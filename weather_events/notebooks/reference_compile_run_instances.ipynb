Reference code to show how to get list of all dates and files that have been ran operationally

```python
import pandas as pd
from pathlib import Path

# 1. Define your starting directory
base_dir = Path("/home/csutter/DRIVE-clean/operational_runs_wMRMS/data_6_ensembling")

# 2. Search for the files
# Using .glob() with this pattern ensures we only look exactly 4 folders deep 
# (Year/Month/Day/Datetime_Folder) for files named 'finalpreds.csv'
file_paths = list(base_dir.glob('*/*/*/*/finalpreds.csv'))

# 3. Extract the data into a list of dictionaries
data = []
for path in file_paths:
    # path.parent.name automatically grabs the name of the folder the file lives in (e.g., "20220112_0000")
    datetime_string = path.parent.name 
    
    data.append({
        "datetime": datetime_string,
        "filepath": str(path) # Convert the Path object back to a standard string
    })

# 4. Create the DataFrame
df = pd.DataFrame(data)

# 5. BONUS: Convert that string into a true datetime object so it's ready for plotting/sorting!
# df['datetime_dt'] = pd.to_datetime(df['datetime'], format='%Y%m%d_%H%M')

# Check your work
print(df.head())

print(len(df))
```
