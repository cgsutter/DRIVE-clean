```python
import pandas as pd
from pathlib import Path

# 1. Define your starting directory
base_dir = Path("/home/csutter/DRIVE-clean/operational_runs_wMRMS/data_6_ensembling")

# 2. Search for the files
# Using .glob() with this pattern ensures we only look exactly 4 folders deep 
# (Year/Month/Day/Datetime_Folder) for files named 'finalpreds.csv'
file_paths = list(base_dir.glob('*/*/*/*/finalpreds.csv'))

# 3. Extract the data into a list of dictionaries
data = []
for path in file_paths:
    # path.parent.name automatically grabs the name of the folder the file lives in (e.g., "20220112_0000")
    datetime_string = path.parent.name 
    
    data.append({
        "datetime": datetime_string,
        "filepath": str(path) # Convert the Path object back to a standard string
    })

# 4. Create the DataFrame
df = pd.DataFrame(data)

# 5. BONUS: Convert that string into a true datetime object so it's ready for plotting/sorting!
# df['datetime_dt'] = pd.to_datetime(df['datetime'], format='%Y%m%d_%H%M')

# Check your work
print(df.head())

print(len(df))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors
import imageio.v2 as imageio 
import os
import requests
import gzip
import xarray as xr
from glob import glob
import numpy as np
import matplotlib.path as mpath

gif_path = "/home/csutter/DRIVE-clean/weather_events/data/statewide_gifs/jan2026_storm_evolution_fulldetails.gif" # HERE!! 

dfuse = jan2026_full # HERE!! 
# jan2026
# jan2026_25th_26th

# --- APPLY FONT STYLING ---
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['STIXGeneral', 'Times New Roman', 'DejaVu Serif']

# --- CREATE A PUNCHY RADAR COLORMAP ---
# Default 'Blues' starts at white. We truncate it so the lowest value is a solid light blue.
base_cmap = plt.cm.Blues
punchy_blues = mcolors.LinearSegmentedColormap.from_list(
    'punchy_blues', base_cmap(np.linspace(0.3, 1.0, 256))
)

custom_colors = {
    'dry': '#71c249',           
    'wet': '#70a8db',           
    'snow': 'pink',          
    'snow_severe': '#eb3324',   
    'poor_viz': '#4a148c'       
}

# Load the pre-baked CSVs using standard pandas
ny_border = pd.read_csv("/home/csutter/DRIVE-clean/data_misc/NY_border_coords.csv")
ny_roads = pd.read_csv("/home/csutter/DRIVE-clean/data_misc/NY_roads_coords.csv")

# --- THE FIX: BUILD NY BOUNDARY POLYGONS FOR EXACT FILTERING ---
print("Building NY boundary polygons to filter out out-of-state cameras...")
ny_paths = []
current_poly = []
for lon, lat in zip(ny_border['Longitude'], ny_border['Latitude']):
    if pd.isna(lon):
        if len(current_poly) > 2:
            ny_paths.append(mpath.Path(current_poly))
        current_poly = []
    else:
        current_poly.append((lon, lat))

def get_mrms_s3(ts, product_type, worker_index):
    """Downloads and subsets MRMS data from S3 using pure xarray."""
    date_str = ts.strftime('%Y%m%d')
    time_str = ts.strftime('%H%M') + "00"
    
    base_url = f"https://noaa-mrms-pds.s3.amazonaws.com/CONUS/{product_type}_00.00"
    file_name = f"MRMS_{product_type}_00.00_{date_str}-{time_str}.grib2.gz"
    full_url = f"{base_url}/{date_str}/{file_name}"
    
    r = requests.get(full_url)

    if r.status_code == 200:
        temp_grib = f"temp_{product_type}_{ts.strftime('%Y%m%d%H%M%S')}_idx{worker_index}.grib2" 
        
        with open(temp_grib, "wb") as f:
            f.write(gzip.decompress(r.content))

        # You already have cfgrib working in this environment!
        ds_full = xr.open_dataset(temp_grib, engine="cfgrib", backend_kwargs={'indexpath': ''})
        ds = ds_full.sel(latitude=slice(47.5, 38.5), longitude=slice(278.0, 291.0)).load()
        
        # Convert Longitude from [0 to 360] to [-180 to 180] to match your camera CSVs
        ds = ds.assign_coords(longitude=(((ds.longitude + 180) % 360) - 180))
        ds = ds.sortby('longitude') 
        
        ds_full.close()
        del ds_full
        
        if os.path.exists(temp_grib):
            os.remove(temp_grib)
            for f in glob(f"{temp_grib}*.idx"):
                os.remove(f)
        
        return ds
    else:
        return None

frame_filenames = []
print("Generating map frames in chronological order...")

# 2. THE FLIPBOOK LOOP
for index, row in dfuse.iterrows():
    timestamp = row['datetime']
    filepath = row['filepath']
    
    # A. Parse Time for MRMS (Round down to nearest 2-min)
    raw_ts = pd.to_datetime(timestamp, format="%Y%m%d_%H%M")
    even_min = (raw_ts.minute // 2) * 2
    mrms_ts = raw_ts.replace(minute=even_min, second=0)
    
    # B. Fetch MRMS Data
    qpe_ds = get_mrms_s3(mrms_ts, "PrecipRate", index)
    
    # C. Read Camera Data using pure Pandas
    step_df = pd.read_csv(filepath)

    # --- THE FIX: EXACT STATE BOUNDARY CROP ---
    # Grab all the camera coordinates for this timeframe
    points = step_df[['Longitude', 'Latitude']].values
    in_ny_mask = np.zeros(len(points), dtype=bool)
    
    # Check if each camera lives inside ANY of our NY polygons (Mainland, Long Island, etc.)
    for poly_path in ny_paths:
        in_ny_mask |= poly_path.contains_points(points)
        
    # Drop all CT, NJ, PA, etc. cameras!
    step_df = step_df[in_ny_mask]
    # ------------------------------------------
    
    # D. Setup Figure
    fig, ax = plt.subplots(figsize=(15, 12))
    fig.patch.set_facecolor('white') 
    
    # --- LAYER 1: State Border (Bottom) ---
    ax.plot(ny_border['Longitude'], ny_border['Latitude'], color='#aaaaaa', linewidth=1.5, zorder=1)
        
    # # --- LAYER 2: MRMS QPE ---
    # if qpe_ds is not None:
    #     var_name = list(qpe_ds.data_vars)[0] 
    #     da = qpe_ds[var_name]
        
    #     # Check if there is actually precipitation
    #     if float(da.max()) > 0.05:
    #         da = da.where(da > 0.05)
            
    #         # Draw the radar blobs
    #         qpe_plot = da.plot.contourf(
    #             ax=ax, 
    #             x='longitude', 
    #             y='latitude', 
    #             levels=np.linspace(0.05, 5.0, 15), 
    #             cmap='Blues', 
    #             alpha=0.9,          # Deep blue, highly visible
    #             add_colorbar=False, # Tell xarray NOT to draw its own shifting colorbar
    #             zorder=2            # Sits safely under roads and cameras
    #         )

    # --- LAYER 2: MRMS QPE ---
    if qpe_ds is not None:
        var_name = list(qpe_ds.data_vars)[0] 
        da = qpe_ds[var_name]
        
        if float(da.max()) > 0.05:
            da = da.where(da > 0.05)
            
            qpe_plot = da.plot.contourf(
                ax=ax, 
                x='longitude', 
                y='latitude', 
                levels=np.linspace(0.05, 5.0, 15), 
                cmap=punchy_blues,  # <--- USE THE NEW COLORMAP HERE
                alpha=0.8,          # <--- Opacity looks great at 0.8 with the darker blues
                add_colorbar=False, 
                zorder=2            # Safely underneath cameras!
            )
            
    # THE PRO-TRICK: THE DUMMY COLORBAR (Locked)
    # We define the exact same scale we used in contourf (0.05 to 5.0)
    # norm = mcolors.Normalize(vmin=0.05, vmax=5.0)
    # sm = cm.ScalarMappable(cmap='Blues', norm=norm)
    # sm.set_array([]) 
    
    # THE PRO-TRICK: THE DUMMY COLORBAR (Locked)
    norm = mcolors.Normalize(vmin=0.05, vmax=5.0)
    sm = cm.ScalarMappable(cmap=punchy_blues, norm=norm) # <--- UPDATE IT HERE TOO
    sm.set_array([])

    # We draw this dummy colorbar EVERY frame, so the map size is permanently locked
    cbar = plt.colorbar(sm, ax=ax, orientation='horizontal', pad=0.03, shrink=0.5, aspect=40)
    cbar.set_label('Precipitation Rate (mm/hr)', fontsize=12)
    cbar.ax.tick_params(labelsize=10)

    # --- LAYER 3: Roads ---
    ax.plot(ny_roads['Longitude'], ny_roads['Latitude'], color='black', linewidth=0.3, alpha=0.5, zorder=3)

    # --- LAYER 4: Cameras (Top) ---
    for condition, color_hex in custom_colors.items():
        subset = step_df[step_df['select'] == condition]
        if not subset.empty:
            ax.scatter(
                subset['Longitude'], 
                subset['Latitude'], 
                c=color_hex, 
                s=60, 
                label=condition, 
                edgecolors='black', 
                linewidths=0.5, 
                zorder=4
            )
    
    # G. Formatting and Cropping
    ax.set_axis_off()
    
    # The Magic Math Trick to fix map stretching without geopandas!
    ax.set_aspect(1.0 / np.cos(np.radians(43.0)))
    
    # Hardcode the X/Y bounds to fit New York perfectly
    ax.set_xlim(-80.0, -71.5)
    ax.set_ylim(40.4, 45.1)
    
    plt.title(f"Statewide Model Predictions\nTime: {timestamp}", 
              fontsize=24, fontstyle='normal', pad=5, y=1.0)
    plt.legend(loc='upper left', title="Predicted Condition", fontsize=12, title_fontsize=14)
    
    frame_name = f"temp_frame_{timestamp}.png"
    plt.savefig(frame_name, dpi=150, facecolor='white', bbox_inches='tight')
    plt.close(fig) 
    
    frame_filenames.append(frame_name)
    print(f"Saved frame for {timestamp}...")


# 3. STITCH THE FRAMES INTO A GIF
print("\nStitching frames into a GIF...")

with imageio.get_writer(gif_path, mode='I', duration=500, loop=0) as writer:
    for filename in frame_filenames:
        image = imageio.imread(filename)
        writer.append_data(image)

for filename in frame_filenames:
    os.remove(filename)

print(f"Success! Movie saved as: {gif_path}")
```

```python
# [ Don't need to run again ] Saving out NYS boundary and roads info in new format b/c Geopandas and xarray don't work well together

import geopandas as gpd
import pandas as pd
import numpy as np

print("Loading shapefiles...")
land_only = gpd.read_file("/home/csutter/DRIVE-clean/data_misc/cb_2023_us_state_500k.zip")
roads = gpd.read_file("/home/csutter/DRIVE-clean/data_misc/tl_2023_36_prisecroads.zip")

# Project them to standard GPS coordinates so they match the MRMS data natively
ny_land = land_only[land_only['STUSPS'] == 'NY'].to_crs(epsg=4326)
roads_flat = roads.to_crs(epsg=4326)

def extract_coords(gdf):
    """Flattens complex shapefiles into a standard list of X/Y coordinates."""
    x, y = [], []
    for geom in gdf.geometry:
        # Grab all the sub-geometries (islands, individual road segments)
        geoms = geom.geoms if hasattr(geom, 'geoms') else [geom]
        for sub_geom in geoms:
            # Polygons use .exterior.coords, Lines use .coords
            coords = sub_geom.exterior.coords if hasattr(sub_geom, 'exterior') else sub_geom.coords
            coords_array = np.array(coords)
            
            x.extend(coords_array[:, 0])
            y.extend(coords_array[:, 1])
            
            # The Magic Trick: Add a blank space so Matplotlib doesn't draw lines connecting every road to each other!
            x.append(np.nan)
            y.append(np.nan)
            
    return pd.DataFrame({'Longitude': x, 'Latitude': y})

print("Extracting coordinates (this might take a minute for the roads)...")
ny_coords = extract_coords(ny_land)
ny_coords.to_csv("/home/csutter/DRIVE-clean/data_misc/NY_border_coords.csv", index=False)

road_coords = extract_coords(roads_flat)
road_coords.to_csv("/home/csutter/DRIVE-clean/data_misc/NY_roads_coords.csv", index=False)

print("Success! You can now plot these using pure Pandas!")
```
